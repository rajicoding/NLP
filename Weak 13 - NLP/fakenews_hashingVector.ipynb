```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics
```

```python
df=pd.read_csv("news.csv", index_col=None)
```

```python
df
```

```python
dataset=df.drop("Unnamed: 0", axis=1)
```

```python
dataset
```

```python
y=dataset["label"]
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(dataset['text'],y,test_size=0.33,random_state=53)
```

```python
# here for hashingVectorizer we don't need to do the fit_transform, as it can map the text directly to vector using hashing function so transform can be used 
hash_vectorizer=HashingVectorizer(stop_words='english',alternate_sign=False)
hash_X_train=hash_vectorizer.transform(X_train)
hash_X_test=hash_vectorizer.transform(X_test)
print(hash_X_train)
```

```python
# model creation
clf=MultinomialNB()
clf.fit(hash_X_train,Y_train)
pred=clf.predict(hash_X_test)
score=metrics.accuracy_score(Y_test,pred)
print("accuracy:%0.3f"%score)
cm=metrics.confusion_matrix(Y_test,pred,labels=["FAKE","REAL"])
```

```python
print(cm)
```

```python
from sklearn.metrics import classification_report
report=classification_report(Y_test,pred)
print(report)
```
