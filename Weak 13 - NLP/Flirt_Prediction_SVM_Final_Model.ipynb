```python
# checking for missing values, duplicates 
import pandas as pd

print(df_final.shape)
print(df_final.columns.tolist())
print(df_final.isnull().sum())

# Remove rows with missing text or labels
df_final = df_final.dropna(subset=["message", "label"]).copy()

# Standardize message whitespace
df_final["message"] = (
    df_final["message"]
    .astype(str)
    .str.strip()
)

df_final = df_final[df_final["message"] != ""]

# Check duplicate messages
print("Duplicate messages:", df_final["message"].duplicated().sum())
```

```python
# Keep the first message and delete the other duplicate
df_final = df_final.drop_duplicates(
    subset=["message"],
    keep="first"
).reset_index(drop=True)
```

```python
df_final["label"].value_counts()
```

```python
# split the data into training and test data
from sklearn.model_selection import train_test_split
X=df_final["message"]
Y=df_final["label"].astype(int)
X_train,X_test,Y_train,Y_test=train_test_split(X, Y, test_size=0.20, random_state=42, stratify=Y)
print("Training samples:", len(X_train))
print("Testing samples", len(X_test))
print("Training level distribution:")
print(Y_train.value_counts())
```

```python
# Training the dataset with TFIDF and SVM
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(stop_words='english')
X_train_tfidf=tfidf.fit_transform(X_train)
X_test_tfidf=tfidf.transform(X_test)
print(X_train_tfidf)
```

```python
# As per the above comparison Linear SVM is the best_model 
from sklearn.svm import LinearSVC
clf=LinearSVC()
clf.fit(X_train_tfidf,Y_train)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
y_pred=clf.predict(X_test_tfidf)
accuracy = accuracy_score(y_pred,Y_test)
print("Accuracy:",accuracy)
cm=confusion_matrix(y_pred,Y_test)
print("confusion matrix:\n",cm)
classification_rpt=classification_report(y_pred,Y_test,labels=[0,1], target_names=["Non-flirty","flirty"],zero_division=0)
print(classification_rpt)
```

```python
# Merging the continuation messages which are present in different columns into one and keeping it in the chat column
```

```python
import pandas as pd

df1 = df1.copy()

# Identify rows that begin with a date like 10/4/20
is_message_start = pd.to_datetime(
    df1["Date"].astype(str),
    format="%m/%d/%y",
    errors="coerce"
).notna()

merged_rows = []

for _, row in df1.iterrows():
    if is_message_start.loc[row.name]:
        merged_rows.append({
            "Date": row["Date"],
            "Chat": str(row["Chat"]) if pd.notna(row["Chat"]) else "",
            "Time": row["Time"],
            "Name": row["Name"]
        })
    elif merged_rows:
        # Collect text from continuation-row columns in order
        parts = [
            str(row[col]).strip()
            for col in ["Date", "Chat", "Time", "Name"]
            if pd.notna(row[col])
            and str(row[col]).strip().lower() not in ["", "nan", "none"]
        ]

        if parts:
            merged_rows[-1]["Chat"] += "\n" + " ".join(parts)

df1 = pd.DataFrame(merged_rows)
df1 = df1.reset_index(drop=True)

display(df1.head(15))
```

## Sentimental Analysis

```python
def sentimentalAnalysis(data,columnname):
    import matplotlib.pyplot as plt
    import seaborn as sns
    "Importing required Packages"
    from nltk.sentiment.vader import SentimentIntensityAnalyzer
    sid=SentimentIntensityAnalyzer()
    "Deleting null or empty value"
    data.dropna(inplace=True)
    "Creating respective columns"
    data["scores"]=data[columnname].apply(lambda commentText: sid.polarity_scores(commentText))
    data["compound"]=data["scores"].apply(lambda score_dict:score_dict["compound"])
    data["Negative"]=data["scores"].apply(lambda score_dict: score_dict["neg"])
    data["Positive"]=data["scores"].apply(lambda score_dict: score_dict["pos"])
    data["Neutral"]=data["scores"].apply(lambda score_dict:score_dict["neu"])
    "Creating final pos or neg using compound score"
    data["comp_score"]=data["compound"].apply(lambda c: 'pos' if c>0 else 'neg')
    plt.clf()
    comp=sns.countplot(x='comp_score', hue='Name', data=data, palette='magma')
    comp.figure.savefig("flirt_prediction_compscore.png")
    "checking how many pos and neg"
    posneg=pd.DataFrame(data["comp_score"].value_counts())
    return posneg,data
    
```

```python
pos,data_senti=sentimentalAnalysis(df1,columnname="Chat")
```

## Topic Modelling

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
dtm=tfidf.fit_transform(df1['Chat'])
```

```python
from sklearn.decomposition import NMF
nmf_model=NMF(n_components=5, random_state=42)
nmf_model.fit(dtm)
```

```python
for index,topic in enumerate(nmf_model.components_):
    results=([tfidf.get_feature_names_out()[i] for i in topic.argsort()[-10:]])
    print(f"{index}:{results}")
```
