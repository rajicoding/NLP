```python
# split the data into training and test data
from sklearn.model_selection import train_test_split
X=df_final["message"]
Y=df_final["label"].astype(int)
X_train,X_test,Y_train,Y_test=train_test_split(X, Y, test_size=0.20, random_state=42, stratify=Y)
print("Training samples:", len(X_train))
print("Testing samples", len(X_test))
print("Training level distribution:")
print(Y_train.value_counts())
```

```python
# Compare the performance of different models on the created dataframe and displaying the results
```

```python

import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB, ComplementNB
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# 1. Define the classification models
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),
    "Multinomial NB": MultinomialNB(),
    "Complement NB": ComplementNB(),
    "Linear SVM": LinearSVC(
        class_weight="balanced",
        random_state=42,
        max_iter=5000
    )
}

# 2. Train models and collect evaluation results
trained_models = {}
results = []

for name, classifier in models.items():

    # Create TF-IDF + classifier pipeline
    pipeline = Pipeline([
        ("tfidf", TfidfVectorizer(
            ngram_range=(1, 2),
            sublinear_tf=True
        )),
        ("classifier", classifier)
    ])

    # Train the model
    pipeline.fit(X_train, Y_train)

    # Predict test labels
    y_pred = pipeline.predict(X_test)

    # Save the trained pipeline
    trained_models[name] = pipeline

    # Generate label-wise classification metrics
    report = classification_report(
        Y_test,
        y_pred,
        labels=[0, 1],
        target_names=["Non-flirty", "Flirty"],
        output_dict=True,
        zero_division=0
    )

    # Store overall and label-wise metrics
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(Y_test, y_pred),

        "Non-flirty Precision":
            report["Non-flirty"]["precision"],
        "Non-flirty Recall":
            report["Non-flirty"]["recall"],
        "Non-flirty F1":
            report["Non-flirty"]["f1-score"],

        "Flirty Precision":
            report["Flirty"]["precision"],
        "Flirty Recall":
            report["Flirty"]["recall"],
        "Flirty F1":
            report["Flirty"]["f1-score"],

        "Macro F1": report["macro avg"]["f1-score"],
        "Weighted F1": report["weighted avg"]["f1-score"]
    })

# 3. Display the model comparison table
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Flirty F1",
    ascending=False
).reset_index(drop=True)

print("=" * 100)
print("MODEL COMPARISON")
print("=" * 100)
print(results_df.round(3).to_string(index=False))

# 4. Display detailed classification reports and confusion matrices
for name, pipeline in trained_models.items():

    y_pred = pipeline.predict(X_test)

    print("\n" + "=" * 60)
    print(f"MODEL: {name}")
    print("=" * 60)

    print("\nClassification Report:")
    print(classification_report(
        Y_test,
        y_pred,
        labels=[0, 1],
        target_names=["Non-flirty", "Flirty"],
        digits=3,
        zero_division=0
    ))

    print("Confusion Matrix:")
    print(confusion_matrix(Y_test, y_pred, labels=[0, 1]))
```

```python
# Merging the continuation messages which are present in different columns into one and keeping it in the chat column
```

```python
import pandas as pd

df1 = df1.copy()

# Identify rows that begin with a date like 10/4/20
is_message_start = pd.to_datetime(
    df1["Date"].astype(str),
    format="%m/%d/%y",
    errors="coerce"
).notna()

merged_rows = []

for _, row in df1.iterrows():
    if is_message_start.loc[row.name]:
        merged_rows.append({
            "Date": row["Date"],
            "Chat": str(row["Chat"]) if pd.notna(row["Chat"]) else "",
            "Time": row["Time"],
            "Name": row["Name"]
        })
    elif merged_rows:
        # Collect text from continuation-row columns in order
        parts = [
            str(row[col]).strip()
            for col in ["Date", "Chat", "Time", "Name"]
            if pd.notna(row[col])
            and str(row[col]).strip().lower() not in ["", "nan", "none"]
        ]

        if parts:
            merged_rows[-1]["Chat"] += "\n" + " ".join(parts)

df1 = pd.DataFrame(merged_rows)
df1 = df1.reset_index(drop=True)

display(df1.head(15))
```

```python
import pandas as pd
df2=pd.read_csv("hf.txt",header=None,on_bad_lines='skip',encoding='utf8')
df2
```

```python
# Make a working copy to preserve the original dataframe
df_cleaning = df2.copy()

# Combine non-empty cells in each row
df_cleaning["Raw_Text"] = df_cleaning.apply(
    lambda row: " ".join(
        str(value).strip()
        for value in row
        if pd.notna(value) and str(value).strip()
    ),
    axis=1
)

display(df_cleaning[["Raw_Text"]].head(20))
```

```python
import re
import pandas as pd

# Matches WhatsApp message starts like:
# 25/08/2023 9:17 pm - Sender: Message
message_start = re.compile(
    r"^\d{1,2}/\d{1,2}/\d{4}\s+"
    r"\d{1,2}:\d{2}\s*[apAP][mM]\s+-\s+"
)

messages = []
current_message = None

for text in df_cleaning["Raw_Text"].fillna("").astype(str):
    text = text.strip()

    if not text:
        continue

    if message_start.match(text):
        if current_message is not None:
            messages.append(current_message)

        current_message = text
    elif current_message is not None:
        # Append continuation lines to the current message
        current_message += " " + text
    else:
        # Keep track of text before the first detected message
        print("Text before first message:", text[:100])

if current_message is not None:
    messages.append(current_message)

df_parsed = pd.DataFrame({"Raw_Message": messages})

print("Parsed WhatsApp messages:", len(df_parsed))
display(df_parsed.head(20))
```

```python
import re
import pandas as pd

header_pattern = re.compile(
    r"^(?P<Date>\d{1,2}/\d{1,2}/\d{4})\s+"
    r"(?P<Time>\d{1,2}:\d{2}\s*[apAP][mM])\s+-\s+"
    r"(?P<Rest>.*)$"
)

def parse_message(raw_message):
    match = header_pattern.match(raw_message)

    if not match:
        return pd.Series({
            "Date": None,
            "Time": None,
            "Name": None,
            "Chat": raw_message
        })

    rest = match.group("Rest")

    # Split sender and message at the first colon
    if ": " in rest:
        name, chat = rest.split(": ", 1)
    else:
        name, chat = None, rest

    return pd.Series({
        "Date": match.group("Date"),
        "Time": match.group("Time"),
        "Name": name,
        "Chat": chat
    })

df_chat_new = pd.concat(
    [
        df_parsed["Raw_Message"],
        df_parsed["Raw_Message"].apply(parse_message)
    ],
    axis=1
)

display(df_chat_new.head(12))
print("Parsed rows:", len(df_chat_new))
print("Rows missing sender:", df_chat_new["Name"].isna().sum())
```
